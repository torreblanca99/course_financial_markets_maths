# 📒 Notes

## 🧮 Simple Interest

The **future value** $VF$ of money under simple interest is given by:

$$
VF = VP \cdot (1 + r \cdot \frac{d}{360})
$$

Where:
- $VP$ = Present Value  
- $r$ = Annual interest rate  
- $d$ = Days to maturity

The **present value** $VP$ is:

$$
VP = \frac{VF}{1 + r \cdot \frac{d}{360}}
$$

---

## 💵 Zero-Coupon Bond

A **zero-coupon bond** pays no interim interest. Its price $P$ is:

$$
P = \frac{VN}{1 + r \cdot \frac{d}{360}} \tag{02.1}
$$

- $VN$ = Face value of the bond  
- $r$ = Yield  
- $d$ = Days to maturity

Alternatively, using a **discount rate** $dr$:

$$
P = VN \cdot (1 - dr \cdot \frac{d}{360}) \tag{02.2}
$$

Where:

$$
dr = \frac{r}{1 + r \cdot \frac{d}{360}}
$$

---

## 📈 Compound Interest

For $n$ reinvestments every $d$ days:

$$
VF_n = VP \cdot (1 + r \cdot \frac{d}{360})^n
$$

### Continuous Compounding

Using Euler’s number:

$$
VF = VP \cdot e^{rT}
$$

$$
VP = VF \cdot e^{-rT}
$$

Where:
- $T = \frac{d \cdot n}{360}$ is time in years

---

## 🏦 Coupon Bond (Fixed Rate)

A **coupon bond** pays periodic interest (coupons) and returns principal at maturity.

The $j$-th coupon:

$$
c_j = VN \cdot rc \cdot \frac{d_j}{360}
$$

Where:
- $VN$ = Nominal value  
- $rc$ = Coupon rate  
- $d_j$ = Days in period $j$

### Dirty Price (Full Price)

$$
P = \sum_{j=1}^n c_j \cdot F_j + F_n \cdot VN
$$

Where:

$$
F_j = \frac{1}{(1 + r \cdot \frac{d_j}{360})^{j - dc/d_1}}
$$

- $dc$ = Days since last coupon

### Clean Price

$$
P_L = P - c \cdot \frac{dc}{d_1}
$$

---

## ⏩ Forward Rate & FRA (Forward Rate Agreement)

The **forward rate** $f$ from $t_1$ to $t_2$:

$$
f = \frac{1 + r_2 \cdot t_2}{1 + r_1 \cdot t_1} - 1 \tag{03.1}
$$

Also:

$$
f_j = \left( \frac{df_{j-1}}{df_j} - 1 \right) \cdot \frac{360}{d_j - d_{j-1}}
$$

- $df_j = \frac{1}{1 + r_j \cdot t_j}$ = discount factor

### FRA: Forward Rate Agreement

An OTC contract where:
- One party pays fixed rate $r_{fix}$  
- The other pays floating rate $r_{float}$

Cash flow at $t_2$:

$$
P\&L = VN \cdot (r_{float} - r_{fix}) \cdot (t_2 - t_1)
$$

If closed early at $t_1$:

$$
P\&L = \frac{VN \cdot (r_{float} - r_{fix}) \cdot (t_2 - t_1)}{1 + r_{float} \cdot (t_2 - t_1)}
$$

In practice, $r_{fix} = f$, the forward rate.

# 2. Zero-coupon Bond

$$VP = \frac{VF}{(1+r)^{T}}$$

```python
class ZeroCouponBond:
    def __init__(self, face_value, risk_free_rate, time_to_maturity):
        """
        Initializer or constructor.
        """
        self.face_value = face_value
        self.risk_free_rate = risk_free_rate
        self.time_to_maturity = time_to_maturity

    def price(self):
        """
        Compute the price of the zero-coupon bond.

        Returns:
        float: Price.
        """
        return self.face_value / (1 + self.risk_free_rate) ** self.time_to_maturity

```

```python
# Example usage
face_value = 1000
risk_free_rate = 0.03
time_to_maturity = 5

bond = ZeroCouponBond(
    face_value, 
    risk_free_rate, 
    time_to_maturity
)
bond_price = bond.price()
print(f"The price of the zero-coupon bond is: {bond_price:.2f}")
```

# 3. Coupon Bond

$$ P = \sum_{t=1}^{N} \frac{C}{(1+r)^{t}} + \frac{F}{(1+r)^{N}}  $$

```python
class CouponBond:
    def __init__(self, f_v, c_r, y_to_mat, ytm):
        """
        Initializer or constructor.
        """
        self.face_value = f_v
        self.coupon_rate = c_r
        self.years_to_maturity = y_to_mat
        self.ytm = ytm
        self.precio = self.price()

    def price(self):
        """
        Compute the price of the coupon bond.
        Returns:
            float: Price.
        """

        annual_coupon_payment = self.face_value * (self.coupon_rate / 100)
        bond_price = 0
        # Present value of the coupons
        for t in range(1, self.years_to_maturity + 1):
            bond_price += annual_coupon_payment / ((1 + self.ytm / 100) ** t)
        bond_price += self.face_value / ((1 + self.ytm / 100) ** self.years_to_maturity)
        return bond_price
```

```python
# Parámetros
face_value = float(input("Ingrese el valor nominal del bono: "))
coupon_rate = float(input("Ingrese el porcentaje que paga el cupón: "))
years_to_maturity = int(input("Ingrese el tiempo de madurez en años: "))
ytm = float(input("Ingrese la tasa interna de retorno: "))

# Cálculo del precio del bono
price = bond_price(face_value, coupon_rate, years_to_maturity, ytm)

print(f"El precio del bono es: ${price:.2f}")
```
